```python

# Importing libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visualization style
sns.set_theme(style="whitegrid", palette="Set2")

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

```python
# Task 1: Load Titanic dataset

df = pd.read_csv('Titanic-Dataset.csv')

# Display dataset dimensions
print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

# Display column names
print("\nDataset Columns:")
print(df.columns.tolist())
```

```python
# Task 2: Inspect the dataset

# Display first five records
print("First five rows:")
display(df.head())

# Display last five records
print("Last five rows:")
display(df.tail())

# Display column data types
print("Data types:")
display(df.dtypes)

# Display descriptive statistics
print("Descriptive statistics:")
display(df.describe())
```

```python

# Check dataset structure
df.info()

# Check missing values
print("\nMissing values:")
display(df.isnull().sum())

# Check duplicate records
print("\nDuplicate records:", df.duplicated().sum())
```

```python
# Create PclassCategory

df['PclassCategory'] = df['Pclass'].map({
    1: 'First Class',
    2: 'Second Class',
    3: 'Third Class'
})

display(df[['Pclass', 'PclassCategory']].head())
```

```python
# Create CabinCategory

df['CabinCategory'] = df['Cabin'].apply(
    lambda x: str(x)[0] if pd.notna(x) else 'Unknown'
)

display(df[['Cabin', 'CabinCategory']].head(10))
```

```python
# Checking missing Age values

print("Missing Age values before cleaning:",
      df['Age'].isnull().sum())

# Calculate mean age
mean_age = df['Age'].mean()

print("Mean passenger age:", mean_age)

# Fill missing Age values
df['Age'] = df['Age'].fillna(mean_age)

# Verify missing values
print("Missing Age values after cleaning:",
      df['Age'].isnull().sum())
```

```python
# Creating Family feature

df['Family'] = df['SibSp'] + df['Parch'] + 1

# Display family information
display(df[['Name', 'SibSp', 'Parch', 'Family']].head(10))
```

Data preprocessing improves the quality and usability of the dataset. Missing age values were replaced using the mean age, while categorical features were created to simplify passenger-class and cabin analysis. The Family feature provides additional information about the total number of people in each passenger's family group.

```python
# Find passengers with the highest fare

highest_fare = df['Fare'].max()

highest_fare_passengers = df[
    df['Fare'] == highest_fare
]

print("Highest Fare:", highest_fare)

display(
    highest_fare_passengers[
        ['Name', 'Sex', 'Pclass', 'Fare']
    ]
)
```

```python
# Find passengers with the largest family size

largest_family = df['Family'].max()

largest_family_passengers = df[
    df['Family'] == largest_family
]

print("Largest Family Size:", largest_family)

display(
    largest_family_passengers[
        ['Name', 'SibSp', 'Parch', 'Family', 'Survived']
    ]
)
```

The analysis identified three passengers who paid the highest fare of approximately 512.33. The largest family group contained 11 members. These findings demonstrate how Pandas filtering and aggregation can identify specific records and unusual observations in a dataset.

```python
# Creating a Pandas Series

names_series = pd.Series(df['Name'])

print("Passenger Names Series:")
display(names_series.head(10))

# Access individual values
print("First passenger:", names_series.iloc[0])

# Slice the Series
print("\nFirst five passengers:")
display(names_series.iloc[0:5])
```

```python
# Creating a DataFrame

names_ages_df = df[['Name', 'Age']].copy()

names_ages_df['Age'] = names_ages_df['Age'].round(2)

print("Names and Ages DataFrame:")
display(names_ages_df.head(10))
```

```python
# Select the first five rows
display(names_ages_df.iloc[0:5])

# Select passengers older than 50
older_passengers = names_ages_df[
    names_ages_df['Age'] > 50
]

print("Passengers older than 50:")
display(older_passengers.head(10))

# Select specific columns
display(df.loc[0:4, ['Name', 'Age', 'Sex']])
```

Pandas Series and DataFrames provide flexible ways to organize and manipulate information. Indexing and slicing allow specific records to be retrieved, while conditional filtering makes it possible to identify passengers based on characteristics such as age.

```python
# Task 6: Create a Multi-index DataFrame

multi_index_df = df.groupby(
    ['Sex', 'Pclass']
)['Survived'].mean().to_frame(
    name='Survival Rate'
)

print("Survival Rates by Sex and Passenger Class:")

display(multi_index_df)
```

Multi-indexing reveals differences in survival rates across passenger sex and class. Female passengers generally experienced higher survival rates than male passengers, while first-class passengers had better survival outcomes than those in lower classes.

```python
# Histogram of Passenger Ages

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Age',
    bins=30,
    kde=True,
    color='steelblue'
)

plt.title('Distribution of Titanic Passenger Ages',
          fontsize=14)

plt.xlabel('Passenger Age')
plt.ylabel('Number of Passengers')

plt.tight_layout()
plt.show()
```

The histogram shows the distribution of passenger ages. Most passengers were young adults, with the average age approximately 29.7 years.

```python
# Bar plot of Survival Rates by Passenger Class

plt.figure(figsize=(9, 6))

sns.barplot(
    data=df,
    x='Pclass',
    y='Survived',
    hue='Pclass',
    palette='inferno',
    legend=False,
    errorbar=None
)

plt.title('Survival Rate by Passenger Class',
          fontsize=14)

plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')

plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```

First-class passengers had the highest survival rate, while third-class passengers had the lowest.

```python
# Correlation Heatmap

plt.figure(figsize=(12, 8))

numerical_df = df.select_dtypes(
    include=np.number
).drop(columns=['PassengerId'])

correlation_matrix = numerical_df.corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Features',
          fontsize=14)

plt.tight_layout()
plt.show()
```

Interpretation:

The heatmap displays relationships between numerical variables.
- Positive correlation means two variables tend to increase together.
- Negative correlation means one variable tends to decrease as the other increases.
- Values close to zero indicate weak linear relationships.
Correlation does not establish causation

```python
# Pair Plot

sns.pairplot(
    df[['Age', 'Fare', 'SibSp', 'Parch', 'Survived']],
    hue='Survived',
    palette='magma',
    diag_kind='hist'
)

plt.suptitle(
    'Pair Plot of Titanic Passenger Features',
    y=1.02
)

plt.show()
```

The pair plot helps visualize relationships between age, fare, siblings/spouses, and parents/children. Different colours represent passengers who survived and those who did not.

It also helps identify clusters, unusual observations, and potential relationships between variables.

```python
# Survival rates by sex

survival_by_sex = df.groupby(
    'Sex'
)['Survived'].mean() * 100

print("Survival Rates by Sex:")
display(survival_by_sex.round(2))
```

```python

# Survival rates by passenger class

survival_by_class = df.groupby(
    'Pclass'
)['Survived'].mean() * 100

print("Survival Rates by Passenger Class:")
display(survival_by_class.round(2))
```

```python
# Survival rates by sex and passenger class

survival_comparison = df.pivot_table(
    values='Survived',
    index='Sex',
    columns='Pclass',
    aggfunc='mean'
) * 100

plt.figure(figsize=(9, 6))

sns.heatmap(
    survival_comparison,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)

plt.title('Survival Rates by Sex and Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Sex')

plt.tight_layout()
plt.show()
```


The analysis indicates that sex and passenger class were strongly associated with survival outcomes. Female passengers had considerably higher survival rates than male passengers. First-class passengers also experienced better survival outcomes compared with passengers in second and third class.

```python
# Box Plot: Age Distribution by Class and Sex

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='Pclass',
    y='Age',
    hue='Sex',
    palette='Set2'
)

plt.title('Age Distribution by Passenger Class and Sex',
          fontsize=14)

plt.xlabel('Passenger Class')
plt.ylabel('Passenger Age')

plt.tight_layout()
plt.show()
```

```python
# Violin Plot: Fare Distribution

plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x='Pclass',
    y='Fare',
    hue='Survived',
    split=True,
    inner='quart',
    palette='Set1'
)

plt.title('Fare Distribution by Passenger Class and Survival Status',
          fontsize=14)

plt.xlabel('Passenger Class')
plt.ylabel('Ticket Fare')

plt.tight_layout()
plt.show()
```
