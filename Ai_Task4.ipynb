```python
# Importing libraries

import pandas as pd
import numpy as np
from functools import reduce

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 100)

print("Python and Pandas environment initialized successfully.")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# TASK 1: Create a DataFrame

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco',
             'Los Angeles', 'Chicago']
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print("Original DataFrame:")
print(df)

# Check column names and data types
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

# Check DataFrame dimensions
print("\nDataFrame Shape:", df.shape)
```

```python
# TASK 2: Drop the City column

df_without_city = df.drop(columns=['City'])

# Display the modified DataFrame
print("DataFrame after dropping City:")
print(df_without_city)

# Verify remaining columns
print("\nRemaining Columns:")
print(df_without_city.columns.tolist())

# Verify the City column no longer exists
assert 'City' not in df_without_city.columns

print("\nVerification successful!")
print("New DataFrame Shape:", df_without_city.shape)
```

```python
# TASK 3: Handling Null Values

data_null = {
    'Name': ['John', None, 'Mary', 'David'],
    'Age': [25, 30, np.nan, 40],
    'City': ['Toronto', 'Ottawa', None, 'Calgary']
}

df_null = pd.DataFrame(data_null)

# Display original DataFrame
print("Original DataFrame:")
print(df_null)

# Identify missing values
print("\nMissing Values Per Column:")
print(df_null.isnull().sum())

# Fill missing values with appropriate replacements
df_filled = df_null.copy()

df_filled['Name'] = df_filled['Name'].fillna('Unknown')
df_filled['Age'] = df_filled['Age'].fillna(0)
df_filled['City'] = df_filled['City'].fillna('Unknown')

# Display cleaned DataFrame
print("\nDataFrame After Filling Missing Values:")
print(df_filled)

# Confirm no missing values remain
print("\nRemaining Null Values:")
print(df_filled.isnull().sum())
```

```python
# TASK 4: GroupBy and Describe

# Define the missing df_group DataFrame with the data matching the descriptions
df_group = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'C'],
    'Value': [15, 20, 20, 20, 25, 35]
})

# Group data by Category
grouped_data = df_group.groupby('Category')

# Apply describe() to the Value column
statistics = grouped_data['Value'].describe()

# Display results
print("Descriptive Statistics by Category:")
print(statistics)

# Display average value for each category
print("\nAverage Value by Category:")
print(grouped_data['Value'].mean())
```

The results show that Category A has three observations with an average value of approximately 18.33. Category B has two observations with an average of 22.5, while Category C has one observation with a value of 35. Category C has the highest recorded value, although its single observation is insufficient to establish a meaningful pattern.

```python
df1 = pd.DataFrame({
    'A': [1, 2],
    'B': [3, 4]
})

df2 = pd.DataFrame({
    'A': [5, 6],
    'B': [7, 8]
})

df3 = pd.DataFrame({
    'C': [9, 10],
    'D': [11, 12]
})
```

```python
# TASK 5: Concatenation and Merging

# Concatenate df1 and df2 vertically
combined_vertical = pd.concat(
    [df1, df2],
    axis=0,
    ignore_index=True
)

print("After Vertical Concatenation:")
print(combined_vertical)

# Combine the result with df3 horizontally
combined_horizontal = pd.concat(
    [combined_vertical, df3],
    axis=1
)

print("\nAfter Horizontal Concatenation:")
print(combined_horizontal)

# Verify dimensions
print("\nFinal DataFrame Shape:")
print(combined_horizontal.shape)

# Verify all expected columns exist
assert list(combined_horizontal.columns) == ['A', 'B', 'C', 'D']
assert combined_horizontal.shape == (4, 4)

print("\nVerification successful!")
```

```python
# TASK 7: Dictionaries

# Create a dictionary of student scores
students = {
    'Alice': 85,
    'Bob': 78,
    'Charlie': 92
}

print("Original Student Dictionary:")
print(students)

# Update an existing student's score
students['Bob'] = 88

print("\nAfter Updating Bob's Score:")
print(students)

# Add a new student
students['David'] = 90

print("\nAfter Adding David:")
print(students)

# Display individual student scores
print("\nStudent Scores:")

for name, score in students.items():
    print(f"{name}: {score}")
```

```python
# TASK 8: Functions and Lambda

# Define a regular function
def calculate_square(number):
    return number ** 2

# Define a lambda function
square_lambda = lambda number: number ** 2

# Test with different inputs
test_numbers = [4, 7, 10]

print("Using Regular Function:")

for number in test_numbers:
    print(f"Square of {number} = {calculate_square(number)}")

print("\nUsing Lambda Function:")

for number in test_numbers:
    print(f"Square of {number} = {square_lambda(number)}")
```

A regular function is defined using the def keyword and is suitable for reusable operations requiring multiple statements. A lambda function is an anonymous, single-expression function that is useful for short operations.

Both functions produce identical results.

```python
# TASK 9: Iterators and Generators

# Create an iterator class
class EvenNumbers:

    def __init__(self):
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 10:
            raise StopIteration

        number = self.current
        self.current += 2

        return number


# Create an iterator object
even_iterator = EvenNumbers()

print("First Five Even Numbers Using Iterator:")

for number in even_iterator:
    print(number)


# Create a generator function
def even_generator():

    for number in range(0, 10, 2):
        yield number


print("\nFirst Five Even Numbers Using Generator:")

for number in even_generator():
    print(number)
```

An iterator is an object that produces values one at a time using __iter__() and __next__().

A generator is a simpler way of creating an iterator using the yield keyword. Generators produce values on demand, which can reduce memory usage when working with large datasets.

```python
# TASK 10: Map, Reduce, and Filter

numbers = [1, 2, 3, 4, 5]

# Use map() to square each number
squared_numbers = list(
    map(lambda x: x ** 2, numbers)
)

# Use reduce() to calculate the product
product = reduce(
    lambda x, y: x * y,
    numbers,
    1
)

# Use filter() to select even numbers
even_numbers = list(
    filter(lambda x: x % 2 == 0, numbers)
)

# Display results
print("Original Numbers:", numbers)

print("\nSquared Numbers:", squared_numbers)

print("\nProduct of Numbers:", product)

print("\nEven Numbers:", even_numbers)
```

```python
# TASK 11: Object-Oriented Programming

# Define Rectangle class
class Rectangle:

    def __init__(self, length, width):

        # Validate dimensions
        if length <= 0 or width <= 0:
            raise ValueError(
                "Length and width must be positive."
            )

        self.length = length
        self.width = width

    # Calculate area
    def calculate_area(self):
        return self.length * self.width

    # Calculate perimeter
    def calculate_perimeter(self):
        return 2 * (self.length + self.width)


# Create first Rectangle object
rectangle1 = Rectangle(10, 5)

# Create second Rectangle object
rectangle2 = Rectangle(8, 4)


# Display results for first rectangle
print("Rectangle 1:")
print("Length:", rectangle1.length)
print("Width:", rectangle1.width)
print("Area:", rectangle1.calculate_area())
print("Perimeter:", rectangle1.calculate_perimeter())


# Display results for second rectangle
print("\nRectangle 2:")
print("Length:", rectangle2.length)
print("Width:", rectangle2.width)
print("Area:", rectangle2.calculate_area())
print("Perimeter:", rectangle2.calculate_perimeter())
```

```python
# TASK 12: Pandas Data Analysis

# Define the df_employees DataFrame with sample data
df_employees = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'Engineering', 'Engineering', 'HR', 'Marketing'],
    'Salary': [55000, 75000, 80000, 48000, 62000]
})

# Display original employee DataFrame
print("Original Employee Data:")
print(df_employees)


# 1. Calculate average salary by department
average_salary = df_employees.groupby(
    'Department'
)['Salary'].mean()

print("\nAverage Salary by Department:")
print(average_salary)


# 2. Extract employees earning more than 60000
high_salary_employees = df_employees.loc[
    df_employees['Salary'] > 60000,
    'Name'
]

print("\nEmployees Earning More Than $60,000:")
print(high_salary_employees)


# 3. Calculate 10% bonus
df_employees['Bonus'] = (
    df_employees['Salary'] * 0.10
)

print("\nEmployee Data With Bonus:")
print(df_employees)


# 4. Verify calculations
assert 'Bonus' in df_employees.columns
assert (df_employees['Bonus'] == df_employees['Salary'] * 0.10).all()

print("\nAll calculations completed successfully!")
```
