```python
import pickle

# 1. Data Types and Variables

book_title = "Python Programming"
book_quantity = 5
book_price = 29.99
is_available = True

print(f"Book: {book_title}, Quantity: {book_quantity}, "
      f"Price: ${book_price}, Available: {is_available}")
```

```python
# 2. Lists

books = [
    "Python Basics",
    "Advanced Python",
    "Data Science with Python"
]

# Add two more books
books.append("Machine Learning Fundamentals")
books.insert(1, "Introduction to Databases")

# Display updated list
print("Updated book list:", books)
```

```python
# 3. Loops and range()

print("\nAvailable books:")

for i in range(len(books)):
    print(f"Index {i}: {books[i]}")
```

```python
# 4. User Input

new_book = input("Enter a new book title: ")

books.append(new_book)

print("Updated book list:", books)
```

```python
# 5. Functions and Conditional Statements

def check_availability(book_name):

    if book_name in books:
        return True
    else:
        return False


# Test the function
print(
    f"Is 'Python Basics' available? "
    f"{check_availability('Python Basics')}"
)

print(
    f"Is 'Computer Networks' available? "
    f"{check_availability('Computer Networks')}"
)
```

```python
def save_books_to_file(filename):
    with open(filename, 'w') as f:
        for book in books:
            f.write(f"{book}\n")
```

```python
# Read books from a text file

def read_books_from_file(filename):

    try:
        with open(filename, 'r') as f:
            book_list = [
                line.strip()
                for line in f
                if line.strip()
            ]

        return book_list

    except FileNotFoundError:
        print("Error: File not found.")
        return []

    except PermissionError:
        print("Error: Permission denied.")
        return []

    except OSError as error:
        print("File error:", error)
        return []


# Test file operations

save_books_to_file("books.txt")

loaded_books = read_books_from_file("books.txt")

print("Books read from file:", loaded_books)
```

```python
# 7. Pickle for serialization
import pickle

def save_books_pickle(filename):

    with open(filename, 'wb') as f:
        pickle.dump(books, f)


# Load books using Pickle

def load_books_pickle(filename):

    try:
        with open(filename, 'rb') as f:
            loaded_books = pickle.load(f)

        if not isinstance(loaded_books, list):
            raise ValueError("Invalid book data format.")

        return loaded_books

    except FileNotFoundError:
        print("Error: Pickle file not found.")
        return []

    except (pickle.UnpicklingError, EOFError):
        print("Error: Unable to read Pickle file.")
        return []

    except (ValueError, OSError) as error:
        print("Error:", error)
        return []


# Test Pickle functions

save_books_pickle("books.pkl")

print(
    "Books loaded from pickle:",
    load_books_pickle("books.pkl")
)
```

```python
# 8. SQLite Database Interaction
import sqlite3

def create_books_table():

    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()

    cursor.execute('''
        CREATE TABLE IF NOT EXISTS books (
            id INTEGER PRIMARY KEY,
            title TEXT,
            quantity INTEGER
        )
    ''')

    conn.commit()
    conn.close()


# Add books to the SQLite database

def add_book_to_db(title, quantity):

    if not isinstance(title, str) or not title.strip():
        print("Error: Invalid book title.")
        return

    if isinstance(quantity, bool) or not isinstance(quantity, int) or quantity < 0:
        print("Error: Quantity must be a non-negative integer.")
        return

    try:
        with sqlite3.connect('library.db') as conn:

            cursor = conn.cursor()

            cursor.execute(
                "INSERT INTO books (title, quantity) VALUES (?, ?)",
                (title.strip(), quantity)
            )

        print(f"Successfully added '{title}' to database.")

    except sqlite3.Error as error:
        print("Database error:", error)


# Test database functions

create_books_table()

add_book_to_db("Python for Beginners", 10)
add_book_to_db("Data Analytics", 5)
```

```python
# Display books stored in SQLite
import sqlite3

with sqlite3.connect('library.db') as conn:

    cursor = conn.cursor()

    cursor.execute("SELECT * FROM books")

    records = cursor.fetchall()

    print("\nBooks stored in SQLite database:")

    for record in records:
        print(record)
```

```python
# 9. Web API Interaction (GitHub API)
import requests

def get_python_repos():

    url = "https://api.github.com/search/repositories"

    params = {
        "q": "language:python",
        "sort": "stars",
        "order": "desc"
    }

    response = requests.get(
        url,
        params=params,
        timeout=15
    )

    response.raise_for_status()

    data = response.json()

    return [
        repo['name']
        for repo in data['items'][:5]
    ]


# Test API function

try:
    print(
        "Top 5 Python repositories on GitHub:",
        get_python_repos()
    )

except requests.exceptions.RequestException as error:
    print("GitHub API request failed:", error)
```
